# **1. Equilibrium constant and free energy**

The equilibrium constant helps us understand whether the reaction tends to have a higher concentration of products or reactants at equilibrium. On the other hand, the Gibbs free energy change of the reaction can describe the spontaneity of the reaction. It is not surprising that they are related. For a general reaction, we have
\begin{equation}
K = -RT\ln(\Delta G^0_r)
\end{equation}
where $K$ is the equilibrium constant, $\Delta G^0_r$ is the change in Gibbs free energy of the reaction, $R$ is the ideal gas constant, and $T$ is the temperature. In Physical Chemistry, we will learn how this equation can be derived.

## **In-class exercise: calculation of equilibrium constant**

Let us consider the reaction $ N_2(g) + O_2(g) ⟺ 2NO(g) $.

At T = 298.15 K, and 1 bar (0.987 atm), the Gibbs free energy change of the reaction is 173.2 kJ/mol. Write a function `equilibriumConst(T)` that outputs the equilibrium constant of the reaction as a function of $T$.



```python
def equilibriumConst(T):

  # Prameters
  R = 8.314               #[J/(mol*K)]
  delta_g_0 = 173.2              #[kJ/mol]
  delta_g_01 = delta_g_0*1000    #[J/mol]

  # Equilibrium constant calculus
  K = np.exp((-delta_g_01)/(R*T))

  return K


print( f"The equilibrium constant at  298.15 K is {equilibriumConst(298.15):.1e}")
```

# **2. Calculation of equilibrium constant**

The composition of the reactants and the products at equilibrium is quantitatively described by the equilibrium constant. For example, for the reaction
\begin{equation}
\text{N}_2(g) + \text{O}_2(g) \rightleftharpoons 2\text{NO}(g)
\end{equation}
the equilibrium constant expression is:
\begin{equation}
K_c = \frac{[\text{NO}]^2}{[\text{N}_2][\text{O}_2]}
\end{equation}
Note that the concentration of each species is raised to the power of its respective coefficient in the balanced chemical equation.

So, if we know the equilibrium constant of a reaction, finding the equilibrium composition is essential about solving the above equation. Let's first take a look at how to solve an equation in Python.

## **Demonstration: Using `fsolve` to Solve an Equation with One Variable**

In this example, we demonstrate how to use the `fsolve` function from the `scipy.optimize` module in Python to solve an equation with a single variable. Specifically, we solve the equation:

\begin{equation}
x^2 - 4 = 0
\end{equation}

This equation has two solutions: \( x = 2 \) and \( x = -2 \). We use the `fsolve` function to numerically find the solution, providing an initial guess to guide the algorithm toward a root of the equation.

The Python code below shows how to define the equation and apply `fsolve` to find the solution:



```python
import numpy as np
from scipy.optimize import fsolve

# Define the equation: x^2 - 4 = 0
def equation(x):
    return x**2 - 4

# Use fsolve to find the solution
initial_guess = 1  # Initial guess for the solution
solution = fsolve(equation, initial_guess)

print(f"The solution to the equation is: x = {solution[0]}")
```

## **In-class exercise: computing NO composition**

In this exercise, let us build a function that outputs the NO concentration as a function of $T$.

We can assume the following approximate composition of the atmospheric air:  $N_2$ = 0.78 e $O_2$ = 0.21.

To get started, we can first build a table that describes the chemical equilibrium and solve the second-degree equation resulted. The equilibrium composition will be calculated by the minimization of the Gibbs free energy considering the equations that describe the process.

|   |   |   |   |
|---|---|---|---|
|  |O2 |N2 |2NO|
|ini |0.21| 0.78 | 0
|eq | 0.21-x | 0.78-x | 2x


Calculate the NO concentration at equilibrium at


1.   298.15 K
2.   1000 K
3.   2000 K

(Note: you may first try to solve the equation at a particular temperature, then turn the code into a function of temperature.)

```python
# [Solution: at a particular temperature]

Temp = 2000                 #[K]

def f(x):
	 return  (2*x)**2 - equilibriumConst(Temp)*(0.78-x)*(0.21-x)

x = fsolve(f,0.01) # find the root of the function. A guess should be informed


c_NO = 2*x


print(f"NO mole fraction at {Temp:.2f} is {c_NO[0]:.1e}")
```

```python
# [Solution: NO concentration as a function of temperature]

def NOconc(Temp):

	def f(x):
		return  (2*x)**2 - equilibriumConst(Temp)*(0.78-x)*(0.21-x)

	x = fsolve(f,0.01) # find the root of the function. A guess should be informed

	c_NO = 2*x

	return c_NO[0]

Temp = 1000
print(f"NO mole fraction at {Temp:.2f} is {NOconc(Temp):.1e}")
```

## **Demonstration: Solving a System of Simultaneous Equations with `fsolve`**

In this example, we demonstrate how to use the `fsolve` function from the `scipy.optimize` module to solve a system of three simultaneous equations with three variables. We will solve the following system of equations:


\begin{aligned}
& \quad x + y + z = 6 \\
& \quad 2x - y + 3z = 14 \\
& \quad 3x + 2y - z = 5
\end{aligned}


We will use `fsolve` to find the values of $x$, $y$, and $z$ that satisfy these equations.


```python
import numpy as np
from scipy.optimize import fsolve

# Define the system of equations
def equations(vars):
    x, y, z = vars
    eq1 = x + y + z - 6
    eq2 = 2*x - y + 3*z - 14
    eq3 = 3*x + 2*y - z - 5
    return [eq1, eq2, eq3]

# Initial guess for the variables (x, y, z)
initial_guess = [1, 1, 1]

# Use fsolve to solve the system of equations
solution = fsolve(equations, initial_guess)

print(f"The solution is: x = {solution[0]}, y = {solution[1]}, z = {solution[2]}")
```

## **In-class exercise: Chemical System with Two Simultaneous Equilibria -- Acid and Water**


A classic example of a chemical system that involves two simultaneous equilibria is the system involving a **weak acid** in water. The two equilibria are:

1. **The dissociation of a weak acid** (e.g., acetic acid, $ \text{CH}_3\text{COOH} $):
   $$
   \text{CH}_3\text{COOH} (aq) \rightleftharpoons \text{H}^+ (aq) + \text{CH}_3\text{COO}^- (aq)
   $$


2. **The dissociation of water** (autoionization of water):
   $$
   \text{H}_2\text{O} (l) \rightleftharpoons \text{H}^+ (aq) + \text{OH}^- (aq)
   $$
  
These two equilibria occur simultaneously in an aqueous solution, and the concentration of $\text{H}^+$ ions will be affected by both the weak acid dissociation and the water dissociation.

Write a program that can compute the pH as a function of the starting acetic acid concentration in a beaker. The **acid dissociation constant** ($K_a$) for acetic acid is $1.8\times 10^{-5}$ and the **water dissociation constant** ($K_w$) is $1.0\times 10^{-14}$ at 25°C.



```python
# [Solution]

import numpy as np
from scipy.optimize import fsolve

# Constants
Ka = 1.8e-5  # Acid dissociation constant for acetic acid
Kw = 1.0e-14  # Water dissociation constant

# Function to compute pH from acetic acid concentration
def compute_pH(c_acetic_acid):
    # Define the system of equations
    def equations(x):
        H_concentration, OH_concentration = x

        # Equation for the dissociation of acetic acid: Ka = [H+][CH3COO-] / [CH3COOH]
        eq1 = Ka - (H_concentration * H_concentration) / (c_acetic_acid - H_concentration)

        # Equation for the dissociation of water: Kw = [H+][OH-]
        eq2 = Kw - (H_concentration * OH_concentration)

        return [eq1, eq2]

    # Initial guess for H+ concentration and OH- concentration
    initial_guess = [1e-7, 1e-7]  # Starting with very low concentrations of H+ and OH-

    # Solve the system of equations using fsolve
    solution = fsolve(equations, initial_guess)

    # The H+ concentration from the solution
    H_concentration = solution[0]

    # Calculate pH
    pH = -np.log10(H_concentration)

    return pH

# Example: Compute pH for different concentrations of acetic acid
acetic_acid_concentrations = [0.1, 0.01, 0.001]  # Example concentrations in Molarity

for c_acetic_acid in acetic_acid_concentrations:
    pH = compute_pH(c_acetic_acid)
    print(f"pH for acetic acid concentration {c_acetic_acid} M: {pH:.2f}")
```
