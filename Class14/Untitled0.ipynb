```python
import psi4

molecule = psi4.geometry("""pubchem:caffeine""")

psi4.energy('scf/3-21g')
```

```python
import py3Dmol

# Get the geometry as a string in XYZ format
xyz_string = molecule.save_string_xyz_file()

# Create a py3dmol view
view = py3Dmol.view(width=400, height=300)

# Add the molecule to the view
view.addModel(xyz_string, 'xyz')

# Set style of the molecule [Try the different styles!]
#view.setStyle({'stick': {}})
view.setStyle({'stick': {'radius': .1}, 'sphere': {'scale': 0.25}})
#view.setStyle({'stick': {'radius': 0.3, 'opacity': 0.5, 'colorscheme':'cyanCarbon'}})
# view.setStyle({'line': {'linewidth':80}})

# Set the camera position
view.zoomTo()

# Set background color
view.setBackgroundColor('white')

# Render the molecule
view.show()
```
